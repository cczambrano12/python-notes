# 05 · Funciones

Una **función** es un bloque de código con nombre que se puede reutilizar. Evita repetir código.

## Definir e invocar

- `def nombre(parámetros):` define la función.
- `return` devuelve un resultado.
- `nombre(argumentos)` la invoca (la ejecuta).

```python
def saludar():
    print("¡Hola!")

saludar()
saludar()
```

## Parámetros y valor de retorno

- **Parámetro:** variable en la definición.
- **Argumento:** valor concreto al invocar.

```python
def sumar(a, b):          # a y b son parámetros
    return a + b

resultado = sumar(3, 4)   # 3 y 4 son argumentos
print(resultado)
```

### `return` vs `print`

`print` solo muestra en pantalla. `return` entrega un valor que se puede usar después.

```python
def doble_print(n):
    print(n * 2)

def doble_return(n):
    return n * 2

x = doble_print(5)    # muestra 10, pero x vale None
y = doble_return(5)   # y vale 10
print(x, y)
print(doble_return(doble_return(3)))   # se puede encadenar
```

Una función sin `return` devuelve `None`. Al ejecutar `return` la función termina.

```python
def es_par(n):
    if n % 2 == 0:
        return True
    return False

print(es_par(4), es_par(7))
```

## Tipos de parámetros

### Posicionales

Importa el orden.

```python
def presentar(nombre, edad):
    print(f"{nombre} tiene {edad} años")

presentar("Luis", 30)
presentar(30, "Luis")   # orden incorrecto, resultado absurdo
```

### Por nombre (keyword)

El orden ya no importa.

```python
presentar(edad=30, nombre="Luis")
```

### Con valor por defecto (opcionales)

Si no se pasa el argumento, se usa el valor por defecto. Los opcionales van **al final**.

```python
def saludar(nombre, saludo="Hola"):
    print(f"{saludo}, {nombre}")

saludar("Ana")
saludar("Ana", "Buenos días")
saludar("Ana", saludo="Qué tal")
```

```python
def calcular_precio(base, iva=0.16, descuento=0):
    return base * (1 + iva) * (1 - descuento)

print(calcular_precio(100))
print(calcular_precio(100, descuento=0.1))
print(calcular_precio(100, iva=0))
```

## Alcance de variables (scope)

Una variable creada dentro de una función **solo existe dentro** de ella.

```python
def calcular():
    resultado = 42   # variable local
    return resultado

print(calcular())
# print(resultado)   # <- NameError: no existe fuera de la función
```

## Documentar una función: docstring

Un **docstring** es un texto entre triples comillas `"""` justo debajo de la línea `def`. Describe qué hace la función, qué recibe y qué devuelve.

```python
def area_rectangulo(base, altura):
    """Calcula el área de un rectángulo.

    Parámetros:
        base: largo de la base.
        altura: largo de la altura.

    Devuelve:
        El área como número.
    """
    return base * altura

print(area_rectangulo(3, 4))
```

`help(función)` muestra ese texto. Así se consulta la documentación de cualquier función, incluidas las que trae Python:

```python
help(area_rectangulo)
```

```python
help(len)
```

## Resumen

- `def` define, `return` devuelve, `nombre()` invoca.
- Parámetros: posicionales, por nombre, con valor por defecto.
- Docstring (`"""..."""`) + `help()` para documentar y consultar.
- Las variables locales no se ven fuera de la función.
