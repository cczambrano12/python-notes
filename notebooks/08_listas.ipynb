# 08 · Listas

Una **lista** es una colección **ordenada** y **modificable** de elementos. Se escribe con corchetes `[]`.

## Crear listas

```python
numeros = [10, 20, 30, 40]
mezcla = ["Ana", 25, True, 3.5]   # pueden mezclar tipos
vacia = []
letras = list("hola")             # list() convierte otra cosa en lista
rango = list(range(5))            # [0, 1, 2, 3, 4]

print(numeros, mezcla, vacia, letras, rango, sep="\n")
```

## Indexación y rebanadas

Igual que en las cadenas: empieza en 0, admite negativos y `[inicio:fin:paso]`.

```python
frutas = ["manzana", "pera", "uva", "kiwi", "mango"]

print(frutas[0])
print(frutas[-1])
print(frutas[1:3])
print(frutas[::-1])
print(len(frutas))
```

A diferencia de las cadenas, las listas **sí se pueden modificar**:

```python
frutas[0] = "fresa"
print(frutas)
```

## Agregar elementos

| Método | Efecto |
|--------|--------|
| `append(x)` | agrega `x` al final |
| `insert(i, x)` | inserta `x` en la posición `i` |
| `extend(otra)` | agrega todos los elementos de otra lista |

```python
compras = ["pan"]
compras.append("leche")
compras.insert(0, "huevos")
compras.extend(["queso", "jamón"])
print(compras)
```

## Quitar elementos

| Método | Efecto |
|--------|--------|
| `remove(x)` | quita la primera aparición de `x` |
| `pop(i)` | quita y **devuelve** el elemento en `i` (por defecto el último) |
| `del lista[i]` | elimina por posición |
| `clear()` | vacía la lista |

```python
l = ["a", "b", "c", "d", "e"]
l.remove("b")
ultimo = l.pop()
del l[0]
print(l, ultimo)
```

## Buscar y contar

```python
notas = [7, 9, 6, 9, 10]

print(9 in notas)
print(notas.count(9))
print(notas.index(6))   # posición (error si no existe)
```

## Ordenar e invertir

- `sort()` **modifica** la lista. `sorted(lista)` devuelve una nueva.

```python
nums = [5, 2, 9, 1]

print(sorted(nums))        # nueva lista ordenada; nums no cambia
print(nums)
nums.sort(reverse=True)    # modifica nums
print(nums)
nums.reverse()
print(nums)
```

## Funciones útiles con listas

```python
n = [4, 8, 15, 16, 23, 42]
print(len(n), min(n), max(n), sum(n))
print("Promedio:", sum(n) / len(n))
```

## Recorrer una lista

```python
amigos = ["Luis", "Marta", "Pedro"]

for amigo in amigos:               # por elemento
    print("Hola", amigo)

# enumerate entrega la posición y el elemento en cada vuelta
for posicion, amigo in enumerate(amigos):
    print(posicion, amigo)
```

## Listas por comprensión

Forma compacta de crear listas: `[expresión for x in secuencia if condición]`

```python
cuadrados = [n ** 2 for n in range(1, 6)]
pares = [n for n in range(10) if n % 2 == 0]
mayusculas = [p.upper() for p in ["sol", "luna"]]
print(cuadrados, pares, mayusculas)
```

## Listas como estructuras de datos en programas

Ejemplo: guardar las notas que escribe el usuario y calcular estadísticas.

```python
notas = []
for n in [8, 6, 9, 10, 7]:     # en un programa real: float(input(...))
    notas.append(n)

print("Notas:", notas)
print("Promedio:", sum(notas) / len(notas))
print("Mejor:", max(notas))
aprobadas = [n for n in notas if n >= 7]
print("Aprobadas:", aprobadas)
```

### Pila y cola

- **Pila (stack):** último en entrar, primero en salir → `append` y `pop()`.
- **Cola (queue):** primero en entrar, primero en salir → `append` y `pop(0)`.

```python
pila = []
pila.append("A")
pila.append("B")
pila.append("C")
print(pila.pop(), pila)

cola = ["A", "B", "C"]
print(cola.pop(0), cola)
```

## Resumen

- Listas: ordenadas, modificables, con índices.
- `append`, `insert`, `extend`, `remove`, `pop`, `sort`, `reverse`, `count`, `index`.
- `len`, `min`, `max`, `sum`, `sorted` funcionan con listas.
- Comprensión: `[f(x) for x in datos if cond]`.
