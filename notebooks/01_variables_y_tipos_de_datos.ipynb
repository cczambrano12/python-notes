## Antes de empezar: `print` y comentarios

- `print(...)` **muestra** valores en pantalla.
- Todo lo que va después de `#` es un **comentario**: Python lo ignora y sirve para dejar notas.
- Un **f-string** es un texto con una `f` delante; lo que va entre `{}` se reemplaza por el valor de la variable.

```python
print("Hola, mundo")        # muestra un texto
print(2 + 3)                # muestra el resultado de una operación
print("Suma:", 2 + 3)       # varios valores separados por coma
nombre = "Ana"
print(f"Me llamo {nombre}")  # f-string
```

## Variables

Una **variable** es un nombre que apunta a un valor. Se crea con `=` (asignación).

```python
nombre = "Ana"
edad = 25
altura = 1.68

print(nombre, edad, altura)
```

Una variable puede cambiar de valor (y hasta de tipo):

```python
edad = 25
edad = edad + 1   # toma el valor actual, le suma 1 y lo guarda de nuevo
print(edad)
```

Asignación múltiple:

```python
x, y, z = 1, 2, 3
a = b = 0
print(x, y, z, a, b)

# Intercambiar valores
x, y = y, x
print(x, y)
```

## Tipos de datos básicos

| Tipo | Nombre | Ejemplo | Para qué sirve |
|------|--------|---------|----------------|
| `int` | entero | `42` | contar, enumerar |
| `float` | decimal | `3.14` | medir, calcular |
| `str` | cadena de texto | `"hola"` | texto |
| `bool` | booleano | `True` / `False` | decisiones |
| `NoneType` | nada | `None` | "sin valor" |

`type()` indica el tipo de un valor:

```python
print(type(42))
print(type(3.14))
print(type("hola"))
print(type(True))
print(type(None))
```

## Conversión de tipos (casting)

Se usa el nombre del tipo como función: `int()`, `float()`, `str()`, `bool()`.

```python
print(int("123") + 1)      # texto -> entero
print(float("2.5") * 2)    # texto -> decimal
print(str(99) + " puntos") # número -> texto
print(int(3.9))            # decimal -> entero (corta, no redondea)
print(bool(0), bool(7), bool(""), bool("a"))  # 0, "" y None son False
```

Convertir algo inválido produce un **error**:

```python
# int("hola")   # <- quita el '#' y ejecuta: ValueError
```

## Resumen

- `variable = valor` guarda un valor.
- Tipos básicos: `int`, `float`, `str`, `bool`, `None`.
- `type(x)` consulta el tipo; `int()`, `float()`, `str()` convierten.
