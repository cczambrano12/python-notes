# 12 · Estructuras de datos anidadas

Listas y diccionarios pueden contener otras listas y diccionarios. Así se representan datos reales.

## Lista de listas

```python
tabla = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9],
]
print(tabla[0])      # primera fila
print(tabla[1][2])   # fila 1, columna 2 -> 6

for fila in tabla:
    for valor in fila:
        print(valor, end=" ")
    print()
```

## Lista de diccionarios

La estructura más común para representar **registros** (como filas de una tabla).

```python
estudiantes = [
    {"nombre": "Ana",   "edad": 20, "nota": 9.0},
    {"nombre": "Luis",  "edad": 22, "nota": 7.5},
    {"nombre": "Marta", "edad": 21, "nota": 8.2},
]

print(estudiantes[1]["nombre"])

for e in estudiantes:
    print(e["nombre"], "→", e["nota"])
```

```python
# Operaciones típicas
suma = 0
for e in estudiantes:
    suma += e["nota"]
print("Promedio:", round(suma / len(estudiantes), 2))

# lista por comprensión: solo los nombres de quienes tienen nota >= 8
aprobados = [e["nombre"] for e in estudiantes if e["nota"] >= 8]
print("Notas ≥ 8:", aprobados)

# buscar el mejor recorriendo la lista
mejor = estudiantes[0]
for e in estudiantes:
    if e["nota"] > mejor["nota"]:
        mejor = e
print("Mejor:", mejor["nombre"])
```

```python
# Agregar un registro nuevo
estudiantes.append({"nombre": "Pedro", "edad": 23, "nota": 6.9})
print(len(estudiantes))
```

## Diccionario de listas

Una llave con **varios** valores.

```python
notas = {
    "Ana":  [9, 8, 10],
    "Luis": [7, 6, 8],
}
notas["Luis"].append(9)

for nombre, lista in notas.items():
    print(nombre, lista, "→ promedio", sum(lista) / len(lista))
```

### Agrupar datos

Ejemplo: agrupar palabras por su primera letra.

```python
palabras = ["manzana", "mango", "pera", "piña", "uva"]
grupos = {}
for p in palabras:
    # setdefault: si la letra no existe como llave, la crea con una lista vacía
    grupos.setdefault(p[0], []).append(p)
print(grupos)
```

## Diccionario de diccionarios

```python
usuarios = {
    "ana":  {"edad": 28, "ciudad": "Lima"},
    "luis": {"edad": 31, "ciudad": "Quito"},
}
print(usuarios["luis"]["ciudad"])

for usuario, datos in usuarios.items():
    print(usuario)
    for campo, valor in datos.items():
        print(f"   {campo}: {valor}")
```

## Estructuras más profundas (tipo JSON)

```python
curso = {
    "nombre": "Python 101",
    "profesor": {"nombre": "Carla", "contacto": ["c@ej.com", "555-0000"]},
    "alumnos": [
        {"nombre": "Ana",  "notas": [9, 8]},
        {"nombre": "Luis", "notas": [7, 9]},
    ],
}

print(curso["profesor"]["contacto"][0])
print(curso["alumnos"][1]["notas"][-1])

for alumno in curso["alumnos"]:
    prom = sum(alumno["notas"]) / len(alumno["notas"])
    print(alumno["nombre"], prom)
```

**Regla:** lee de izquierda a derecha, un nivel por corchete: `curso["alumnos"][1]["notas"][-1]` → dict → lista → dict → lista → elemento.


Estas estructuras equivalen al formato **JSON**, muy usado en internet. En el cuaderno `14` verás cómo guardarlas en un archivo.

## Resumen

- Lista de dicts: tabla de registros.
- Dict de listas: varios valores por llave.
- Dict de dicts: registros indexados por nombre.
- Se accede encadenando `[...]` un nivel a la vez.
