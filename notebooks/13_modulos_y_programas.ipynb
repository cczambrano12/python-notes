## Crear y ejecutar un programa de Python

Un **programa** es un archivo de texto con extensión `.py`. Se ejecuta desde una terminal:

```bash
python mi_programa.py
```

Desde Jupyter podemos crear el archivo con `%%writefile` y ejecutarlo con `!python`.

````python
%%writefile programa_basico.py
# Programa que saluda al usuario
nombre = "mundo"
print(f"Hola, {nombre}")
````

````python
!python programa_basico.py
````

### El patrón `if __name__ == "__main__":`

Cuando un archivo se **ejecuta** directamente, `__name__` vale `"__main__"`. Cuando se **importa**, vale el nombre del módulo. Así el archivo puede ser programa y módulo a la vez.

## Módulos propios

Un **módulo** es un archivo `.py` con funciones y variables que otros archivos pueden **importar**. Sirve para separar la lógica en partes pequeñas y reutilizables.

Creamos un módulo `geometria.py`:

````python
%%writefile geometria.py
"""Funciones de geometría básica."""

PI = 3.14159

def area_circulo(radio):
    return PI * radio ** 2

def area_rectangulo(base, altura):
    return base * altura

if __name__ == "__main__":
    # solo se ejecuta si corres: python geometria.py
    print("Prueba:", area_circulo(1))
````

Y un programa principal que lo usa:

````python
%%writefile main.py
import geometria

radio = 2
print("Área del círculo:", geometria.area_circulo(radio))
print("Área del rectángulo:", geometria.area_rectangulo(3, 4))
````

````python
!python main.py
````

### Formas de importar

| Forma | Uso |
|-------|-----|
| `import geometria` | `geometria.area_circulo(2)` |
| `import geometria as g` | `g.area_circulo(2)` (alias) |
| `from geometria import area_circulo` | `area_circulo(2)` |
| `from geometria import *` | evitar: mezcla nombres |

````python
import geometria as g
from geometria import area_rectangulo

print(g.PI)
print(g.area_circulo(1))
print(area_rectangulo(2, 5))
````

Python busca los módulos en la carpeta actual y en las bibliotecas instaladas. Los archivos que creaste están en la misma carpeta que este cuaderno.

## Biblioteca estándar

Python incluye muchos módulos listos para usar, sin instalar nada.

### `math`

````python
import math

print(math.pi, math.e)
print(math.sqrt(16))
print(math.pow(2, 10))
print(math.ceil(4.1), math.floor(4.9))   # redondear hacia arriba / abajo
print(math.factorial(5))
print(math.gcd(12, 18))                  # máximo común divisor
print(math.sin(math.pi / 2), math.cos(0))
print(math.degrees(math.pi), math.radians(180))
print(math.log(100, 10), math.log10(1000), math.log(math.e))
print(math.hypot(3, 4))                  # hipotenusa
````

Para ver todas las funciones: `dir(math)`; para la ayuda de una: `help(math.sqrt)`.

### `random`

````python
import random

print(random.random())            # decimal entre 0 y 1
print(random.randint(1, 6))       # entero entre 1 y 6 (ambos incluidos)
print(random.uniform(1.5, 3.5))   # decimal entre dos valores
print(random.choice(["piedra", "papel", "tijera"]))
````

````python
cartas = ["A", "K", "Q", "J", "10"]
random.shuffle(cartas)            # mezcla la lista (en el sitio)
print(cartas)
print(random.sample(range(1, 50), 6))   # 6 números sin repetir
````

**Semilla:** con `random.seed(n)` los números "aleatorios" se repiten, útil para reproducir resultados.

````python
random.seed(42)
print(random.randint(1, 100), random.randint(1, 100))
random.seed(42)
print(random.randint(1, 100), random.randint(1, 100))
````

### Mini juego: adivina el número

````python
import random

secreto = random.randint(1, 20)
intentos = 0
while True:
    texto = input("Adivina (1-20): ")
    if not texto.isdigit():
        print("Escribe un número.")
        continue
    intentos += 1
    n = int(texto)
    if n < secreto:
        print("Más alto")
    elif n > secreto:
        print("Más bajo")
    else:
        print(f"¡Correcto en {intentos} intentos!")
        break
````

### 🎛️ Dados

Pulsa el botón para lanzar dados y observa cómo se distribuyen.

````python
import ipywidgets as widgets
import random

n_dados = widgets.IntSlider(value=2, min=1, max=6, description="Dados")
lanzar = widgets.Button(description="🎲 Lanzar", button_style="primary")
historial = []
out = widgets.Output()

def on_click(_):
    tirada = [random.randint(1, 6) for _ in range(n_dados.value)]
    historial.append(sum(tirada))
    out.clear_output()
    with out:
        print("Resultado:", tirada, "→ suma", sum(tirada))
        conteo = {}
        for s in historial:
            conteo[s] = conteo.get(s, 0) + 1
        for suma in sorted(conteo):
            print(f"{suma:2} {'█' * conteo[suma]}")

lanzar.on_click(on_click)
display(widgets.HBox([n_dados, lanzar]), out)
````

## Resumen

- Un programa es un `.py`; se ejecuta con `python archivo.py`.
- Un módulo es un `.py` que se importa: `import modulo`, `from modulo import funcion`.
- `if __name__ == "__main__":` separa "ejecutar" de "importar".
- Biblioteca estándar: `math`, `random`, `datetime`, `json`...
