# 02 · Operadores

Los **operadores** combinan valores para producir un resultado. Su comportamiento depende del tipo de dato.

## Números (`int` y `float`)

| Operador | Significado | Ejemplo | Resultado |
|----------|-------------|---------|-----------|
| `+` | suma | `7 + 2` | `9` |
| `-` | resta | `7 - 2` | `5` |
| `*` | multiplicación | `7 * 2` | `14` |
| `/` | división (siempre `float`) | `7 / 2` | `3.5` |
| `//` | división entera | `7 // 2` | `3` |
| `%` | residuo (módulo) | `7 % 2` | `1` |
| `**` | potencia | `7 ** 2` | `49` |

```python
a, b = 7, 2
print(a + b, a - b, a * b)
print(a / b, a // b, a % b)
print(a ** b)
```

**Precedencia:** primero `**`, luego `* / // %`, luego `+ -`. Los paréntesis mandan.

```python
print(2 + 3 * 4)
print((2 + 3) * 4)
print(2 ** 3 ** 2)   # se evalúa de derecha a izquierda: 2 ** 9
```

**Uso típico de `%`:** saber si un número es par o múltiplo de otro.

```python
print(10 % 2 == 0)   # ¿10 es par?
print(15 % 4)        # sobra 3
```

### Operadores de asignación compuesta

```python
puntos = 10
puntos += 5    # puntos = puntos + 5
puntos *= 2    # puntos = puntos * 2
puntos -= 1
print(puntos)
```

### Precisión de los decimales

Los `float` se guardan en binario y pueden tener pequeños errores:

```python
print(0.1 + 0.1 + 0.1)
print(round(0.1 + 0.1 + 0.1, 2))   # round(valor, decimales)
```

## Texto (`str`)

Solo `+` (unir) y `*` (repetir) tienen sentido con texto.

```python
print("Hola, " + "mundo")
print("ja" * 3)
print("-" * 20)
# print("edad: " + 25)   # <- error: no se puede sumar str con int
print("edad: " + str(25))
```

## Comparaciones

Devuelven `True` o `False`.

| Operador | Significado |
|----------|-------------|
| `==` | igual |
| `!=` | distinto |
| `<` `>` | menor / mayor |
| `<=` `>=` | menor o igual / mayor o igual |

⚠️ `=` asigna, `==` compara.

```python
print(5 == 5, 5 != 5)
print(3 < 7, 3 >= 7)
print("a" < "b")             # el texto se compara alfabéticamente
print("hola" == "Hola")      # mayúsculas importan
print(1 < 5 < 10)            # comparaciones encadenadas
```

## Pertenencia: `in`

Sirve con texto, listas, diccionarios, etc.

```python
print("sol" in "girasol")
print("z" not in "hola")
```

## Listas (adelanto)

Las listas también soportan `+` y `*`:

```python
print([1, 2] + [3])
print([0] * 5)
```

## Ejercicio interactivo: calculadora

Elige dos números y un operador.

```python
import ipywidgets as widgets

n1 = widgets.FloatText(value=7, description="a:", layout=widgets.Layout(width="150px"))
n2 = widgets.FloatText(value=2, description="b:", layout=widgets.Layout(width="150px"))
op = widgets.ToggleButtons(options=["+", "-", "*", "/", "//", "%", "**"], description="Operador:")
res = widgets.Output()

def calcular(_=None):
    res.clear_output()
    with res:
        a, b = n1.value, n2.value
        try:
            r = eval(f"a {op.value} b")
            print(f"{a} {op.value} {b} = {r}")
        except ZeroDivisionError:
            print("¡No se puede dividir entre cero!")

for w in (n1, n2, op):
    w.observe(calcular, names="value")
calcular()
display(widgets.HBox([n1, n2]), op, res)
```

## Resumen

- Aritmética: `+ - * / // % **`.
- Comparación: `== != < > <= >=` → `bool`.
- `+` y `*` también funcionan con texto y listas.
- `in` revisa si algo está contenido en otra cosa.
