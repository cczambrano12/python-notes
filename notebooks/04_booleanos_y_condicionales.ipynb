# 04 · Booleanos y condicionales (`if`, `elif`, `else`)

Los programas toman decisiones según condiciones que son verdaderas (`True`) o falsas (`False`).

## Operadores booleanos

| Operador | Significado | Es `True` cuando... |
|----------|-------------|---------------------|
| `and` | y | **ambas** son verdaderas |
| `or` | o | **al menos una** es verdadera |
| `not` | negación | invierte el valor |

```python
a, b = True, False
print("a and b:", a and b)
print("a or b: ", a or b)
print("not a:  ", not a)
```

### Tablas de verdad

```python
print("True  and True  →", True and True)
print("True  and False →", True and False)
print("False and True  →", False and True)
print("False and False →", False and False)
print()   # línea en blanco
print("True  or True  →", True or True)
print("True  or False →", True or False)
print("False or True  →", False or True)
print("False or False →", False or False)
```

Precedencia: `not` → `and` → `or`. Usa paréntesis para dejarlo claro.

```python
edad = 20
tiene_permiso = False
print(edad >= 18 and not tiene_permiso)
print(edad < 18 or tiene_permiso)
```

## `if`

Ejecuta un bloque **solo si** la condición es verdadera. El bloque se marca con **dos puntos** `:` y **sangría** (4 espacios).

```python
temperatura = 32

if temperatura > 30:
    print("Hace calor")
    print("Toma agua")

print("Esto se imprime siempre")
```

## `if` / `else`

```python
edad = 16

if edad >= 18:
    print("Mayor de edad")
else:
    print("Menor de edad")
```

## `if` / `elif` / `else`

Evalúa las condiciones **en orden** y ejecuta solo la primera que sea verdadera.

```python
nota = 85

if nota >= 90:
    letra = "A"
elif nota >= 80:
    letra = "B"
elif nota >= 70:
    letra = "C"
else:
    letra = "F"

print(f"Nota {nota} → {letra}")
```

⚠️ **El orden importa.** Si pones `nota >= 70` primero, 85 caería ahí y nunca llegaría a `B`.

### `if` anidados y condiciones combinadas

```python
edad = 25
tiene_licencia = True

if edad >= 18 and tiene_licencia:
    print("Puede conducir")
elif edad >= 18:
    print("Debe sacar la licencia")
else:
    print("Aún no puede conducir")
```

### Expresión condicional (una línea)

```python
edad = 20
estado = "adulto" if edad >= 18 else "menor"
print(estado)
```

### Valores "verdaderos" y "falsos"

Son falsos: `0`, `0.0`, `""`, `[]`, `{}`, `None`, `False`. Todo lo demás es verdadero.

```python
nombre = ""
if nombre:
    print("Hola", nombre)
else:
    print("No escribiste tu nombre")
```

## Resumen

- `and`, `or`, `not` combinan condiciones.
- `if` / `elif` / `else` eligen **un** camino; el orden importa.
- La sangría define qué código pertenece al bloque.
