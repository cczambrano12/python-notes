# 09 · Tuplas

Una **tupla** es como una lista pero **inmutable**: no se puede modificar después de crearla. Se escribe con paréntesis `()`.

```python
punto = (3, 4)
persona = ("Ana", 28, "Lima")
una_sola = (5,)      # ¡la coma es lo que hace a la tupla!
vacia = ()
sin_parentesis = 1, 2, 3
sin_parentesis
```

```python
type(una_sola)
```

```python
type((5))
```

## Inmutabilidad

```python
# punto[0] = 10   # <- TypeError: no se puede modificar
punto = (10, punto[1])   # se crea una tupla nueva
punto
```

## Desempaquetado

Asigna cada elemento a una variable en un solo paso.

```python
nombre, edad, ciudad = persona
nombre, edad, ciudad
```

```python
primero, *resto = (1, 2, 3, 4)   # *resto recoge los demás en una lista
primero, resto
```

Las funciones que "devuelven varios valores" devuelven una tupla:

```python
def dividir(a, b):
    return a // b, a % b      # devuelve dos valores

cociente, residuo = dividir(17, 5)
cociente, residuo
```

```python
def min_y_max(a, b, c):
    return min(a, b, c), max(a, b, c)   # min y max: el menor y el mayor

menor, mayor = min_y_max(5, 1, 9)
menor, mayor
```

## Conversión

```python
lista = list((1, 2, 3))
lista
```

```python
tupla = tuple([4, 5, 6])
tupla
```

## ¿Cuándo usar tupla en vez de lista?

- Datos que **no deben cambiar**: coordenadas, fechas `(2025, 12, 31)`, RGB `(255, 0, 0)`.
- Devolver varios valores de una función.
- Son más ligeras y, a diferencia de las listas, se pueden usar como **llaves de diccionario** (cuaderno `10`).

## Resumen

| | Lista | Tupla |
|---|-------|-------|
| Sintaxis | `[1, 2]` | `(1, 2)` |
| Modificable | Sí | No |
| Llave de diccionario (cuaderno 10) | No | Sí |
