# 10 · Diccionarios

Un **diccionario** (`dict`) guarda **pares llave → valor**. Se accede a un valor por su llave, no por una posición. Como en un diccionario de palabras: la palabra es la llave y la definición es el valor.

## Crear diccionarios

```python
persona = {"nombre": "Ana", "edad": 28, "ciudad": "Lima"}
vacio = {}
otro = dict(a=1, b=2)

print(persona)
print(type(persona), len(persona))
```

- Las **llaves** son únicas y deben ser inmutables (`str`, `int`, tuplas...).
- Los **valores** pueden ser de cualquier tipo, incluso otros diccionarios.
- Distinguen mayúsculas: `"Nombre"` ≠ `"nombre"`.

## Acceder a valores

```python
print(persona["nombre"])
# print(persona["telefono"])   # <- KeyError: la llave no existe
print(persona.get("telefono"))               # None, sin error
print(persona.get("telefono", "sin datos"))  # con valor por defecto
```

## ¿Existe la llave?

```python
print("edad" in persona)
print("telefono" not in persona)

if persona.get("telefono") is None:
    print("No hay teléfono registrado")
```

## Agregar, modificar y eliminar

```python
persona["telefono"] = "555-1234"   # agregar
persona["edad"] = 29               # modificar
print(persona)

borrado = persona.pop("telefono")  # elimina y devuelve el valor
del persona["ciudad"]              # elimina
print(persona, borrado)
```

```python
persona.update({"ciudad": "Quito", "pais": "Ecuador"})   # varios a la vez
print(persona)
```

## Métodos principales

| Método | Devuelve |
|--------|----------|
| `d.keys()` | las llaves |
| `d.values()` | los valores |
| `d.items()` | pares `(llave, valor)` |
| `d.get(k, defecto)` | valor o `defecto` |
| `d.pop(k)` | elimina `k` y devuelve su valor |
| `d.update(otro)` | agrega/actualiza desde otro dict |
| `d.setdefault(k, v)` | devuelve `d[k]`; si no existe, lo crea con `v` |
| `d.clear()` | vacía el diccionario |
| `d.copy()` | copia superficial |

```python
notas = {"mate": 9, "historia": 7, "arte": 10}

print(list(notas.keys()))
print(list(notas.values()))
print(list(notas.items()))
```

## Recorrer un diccionario

```python
for materia in notas:                # por llaves
    print(materia)

for nota in notas.values():          # por valores
    print(nota)

for materia, nota in notas.items():  # por ambos (lo más común)
    print(f"{materia}: {nota}")
```

```python
print("Promedio:", sum(notas.values()) / len(notas))
mejor = max(notas, key=notas.get)    # llave con el valor más alto
print("Mejor materia:", mejor)
```

## Aplicaciones

### 1. Histograma (contar ocurrencias)

```python
texto = "murcielago"
frecuencia = {}
for letra in texto:
    frecuencia[letra] = frecuencia.get(letra, 0) + 1

print(frecuencia)
```

### 2. Conjunto de elementos únicos

Usar solo las llaves para saber qué valores aparecieron.

```python
import random
tiradas = [random.randint(1, 6) for _ in range(10)]
vistos = {}
for t in tiradas:
    vistos[t] = True
print(tiradas)
print("Valores distintos:", list(vistos))
```

### 3. Estructura con campos con nombre

Más legible que una lista: `telefono["marca"]` en lugar de `telefono[0]`.

```python
telefono = {"marca": "Acme", "modelo": "X1", "pantalla": 6.1, "ram_gb": 8}
print(f"{telefono['marca']} {telefono['modelo']} — {telefono['ram_gb']} GB")
```

### 4. Tabla de búsqueda (traducción, códigos)

```python
capitales = {"Colombia": "Bogotá", "Perú": "Lima", "Chile": "Santiago"}
pais = "Perú"
print(f"La capital de {pais} es {capitales.get(pais, 'desconocida')}")
```

### 5. Diccionario de diccionarios

```python
telefonos = {
    "A": {"marca": "Acme", "ram_gb": 8},
    "B": {"marca": "Zeta", "ram_gb": 4},
}
print(telefonos["A"]["ram_gb"])   # llaves encadenadas
telefonos["B"]["ram_gb"] = 6
print(telefonos)
```

## Los diccionarios son mutables y se pasan por referencia

Una función puede modificar el diccionario original. `copy()` crea una copia superficial. Más detalle en el cuaderno `11`.

```python
def subir_nota(d):
    d["mate"] += 1

original = {"mate": 5}
subir_nota(original)
print(original)

copia = original.copy()
copia["mate"] = 0
print(original, copia)
```

## 🎛️ Contador de letras

Escribe un texto y mira el histograma.

```python
import ipywidgets as widgets

texto = widgets.Textarea(value="hola mundo", description="Texto:", layout=widgets.Layout(width="400px"))
out = widgets.Output()

def contar(_=None):
    out.clear_output()
    freq = {}
    for c in texto.value.lower():
        if c.isalpha():
            freq[c] = freq.get(c, 0) + 1
    with out:
        for letra, n in sorted(freq.items(), key=lambda p: -p[1]):
            print(f"{letra} {'█' * n} {n}")

texto.observe(contar, names="value")
contar()
display(texto, out)
```

## Resumen

- `{llave: valor}`; acceso con `d[llave]` o `d.get(llave, defecto)`.
- `in` revisa llaves; `keys()`, `values()`, `items()` para recorrer.
- Útiles para contar, buscar, agrupar datos y describir objetos.
