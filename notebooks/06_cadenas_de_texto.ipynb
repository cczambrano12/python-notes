# 06 · Cadenas de texto (`str`)

Una **cadena** es una secuencia de caracteres. Se escribe entre comillas simples `'...'` o dobles `"..."`.

````python
a = "Hola"
b = 'Mundo'
print(a, b)
````

````python
c = """Texto de
varias líneas"""
print(c)
````

## Indexación

Cada carácter tiene una **posición** (índice) que empieza en **0**. Los índices negativos cuentan desde el final.

```
 P  y  t  h  o  n
 0  1  2  3  4  5
-6 -5 -4 -3 -2 -1
```

````python
palabra = "Python"
palabra[0]    # primer carácter
````

````python
palabra[5]    # último
````

````python
palabra[-1]   # último (otra forma)
````

````python
len(palabra)  # len() devuelve la cantidad de caracteres
````

### Rebanadas (slicing): `[inicio:fin:paso]`

Incluye `inicio`, **excluye** `fin`.

````python
palabra[0:2]   # 'Py'
````

````python
palabra[2:]    # 'thon'
````

````python
palabra[:3]    # 'Pyt'
````

````python
palabra[::2]   # 'Pto' (de 2 en 2)
````

````python
palabra[::-1]  # 'nohtyP' (al revés)
````

Las cadenas son **inmutables**: no se pueden modificar por posición.

````python
# palabra[0] = "J"   # <- TypeError
palabra = "J" + palabra[1:]   # se crea una cadena nueva
palabra
````

## Codificación (encoding)

Las computadoras guardan números, no letras. Una **codificación** asigna un número a cada carácter. Python usa **Unicode**; el archivo en disco normalmente usa **UTF-8**.

````python
ord("A")     # carácter -> número
````

````python
ord("ñ")
````

````python
chr(97)      # número -> carácter
````

````python
chr(128512)  # emoji
````

`encode()` convierte texto en **bytes**; `decode()` hace lo inverso.

````python
texto = "canción"
datos = texto.encode("utf-8")
datos             # ñ y ó ocupan 2 bytes cada uno
````

````python
len(texto), len(datos)
````

````python
datos.decode("utf-8")
````

Si decodificas con la codificación equivocada, aparecen caracteres extraños (*mojibake*):

````python
datos.decode("latin-1")
````

## Caracteres de control (escape)

La barra invertida `\` introduce caracteres especiales.

| Secuencia | Significado |
|-----------|-------------|
| `\n` | salto de línea |
| `\t` | tabulador |
| `\\` | barra invertida |
| `\"` `\'` | comillas |
| `\u00f1` | carácter Unicode (ñ) |

````python
print("Línea 1\nLínea 2")
````

````python
print("Col1\tCol2\tCol3")
````

````python
print("Ruta: C:\\Usuarios\\ana")
````

````python
print("Dijo \"hola\"")
````

````python
print("\u00f1 \u2764")
````

**Cadenas crudas** (`r"..."`): la barra no es especial.

````python
print(r"C:\nuevo\tabla")
````

`print` tiene parámetros útiles: `sep` y `end`.

````python
print("a", "b", "c", sep="-")
````

````python
print("sin salto", end=" ")
print("seguido")
````

## Funciones vs. métodos

- **Función:** se llama por sí sola y recibe el dato como argumento → `len(texto)`.
- **Método:** pertenece a un objeto y se llama con punto → `texto.upper()`.

Los métodos de `str` **no modifican** la cadena original: devuelven una nueva.

````python
texto = "hola mundo"
len(texto)      # función
````

````python
texto.upper()   # método
````

````python
texto           # el original no cambió
````

## Métodos de cadenas más usados

| Método | Qué hace |
|--------|----------|
| `upper()` / `lower()` | mayúsculas / minúsculas |
| `capitalize()` / `title()` | primera letra / cada palabra en mayúscula |
| `strip()` | quita espacios de los extremos |
| `replace(a, b)` | reemplaza `a` por `b` |
| `split(sep)` | divide en una lista |
| `sep.join(lista)` | une una lista en una cadena |
| `find(x)` / `index(x)` | posición de `x` (`find` da -1 si no está) |
| `count(x)` | cuántas veces aparece |
| `startswith(x)` / `endswith(x)` | comienza / termina con |
| `isdigit()` / `isalpha()` | solo dígitos / solo letras |

````python
frase = "  Aprender Python es divertido  "
frase.strip()
````

````python
frase.lower()
````

````python
frase.title()
````

````python
frase.replace("divertido", "genial")
````

````python
frase.count("e")
````

````python
frase.find("Python")
````

````python
frase.strip().startswith("Aprender")
````

`split` devuelve una **lista**: varios valores entre corchetes `[...]` (se estudian en el cuaderno `08`). `join` hace lo contrario: une una lista en una sola cadena.

````python
palabras = "manzana,pera,uva".split(",")
palabras
````

````python
" - ".join(palabras)
````

````python
"123".isdigit()
````

````python
"12a".isdigit()
````

````python
"hola".isalpha()
````

````python
"hola mundo".isalpha()
````

## Formato de cadenas

### f-strings (recomendado)

Antepón `f` y escribe variables o expresiones entre `{}`.

````python
nombre = "Ana"
edad = 28
f"{nombre} tiene {edad} años"
````

````python
f"El año que viene tendrá {edad + 1}"
````

````python
f"{nombre=}"   # muestra el nombre y el valor: útil para revisar
````

### Formatear números: `{valor:formato}`

````python
pi = 3.14159265
precio = 1234567.891
fraccion = 0.256
f"{pi:.2f}"          # 2 decimales
````

````python
f"{precio:,.2f}"     # separador de miles
````

````python
f"{fraccion:.1%}"    # porcentaje
````

````python
f"{7:03d}"           # relleno con ceros -> 007
````

````python
f"{'hola':>10}|"     # alineado a la derecha, ancho 10
````

````python
f"{'hola':<10}|"     # izquierda
````

````python
f"{'hola':^10}|"     # centrado
````

### Otra forma (la verás en código antiguo)

````python
"{} tiene {} años".format("Ana", 28)
````

## Resumen

- Índices desde 0; `[inicio:fin:paso]` para rebanar.
- Las cadenas son inmutables; los métodos devuelven cadenas nuevas.
- `\n`, `\t`, `\\` son caracteres de control.
- Usa f-strings: `f"{valor:.2f}"`.
