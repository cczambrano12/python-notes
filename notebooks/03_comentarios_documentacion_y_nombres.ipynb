## Comentarios

Todo lo que va después de `#` es ignorado por Python.

```python
# Esto es un comentario de línea completa
precio = 100
impuesto = 0.16  # comentario al final de la línea

precio * (1 + impuesto)
```

**Buenas prácticas**
- Explica el **por qué**, no el **qué** (el código ya dice qué hace).
- Mantén los comentarios actualizados.
- Si necesitas muchos comentarios, quizá el código se puede simplificar con mejores nombres.

```python
# Mal: repite lo obvio
x = 5  # asigna 5 a x

# Bien: explica la razón
descuento_maximo = 0.30  # política de la tienda: nunca más del 30 %
```

## Documentación

Además de los comentarios con `#`, Python permite documentar las **funciones** con un texto especial llamado *docstring*. Se explica en el cuaderno `05`, justo después de aprender a crear funciones.

## Nombres de variables

### Reglas obligatorias
- Pueden contener letras, números y `_`.
- No pueden empezar con número.
- No pueden tener espacios ni símbolos como `-` o `$`.
- No pueden ser palabras reservadas (`if`, `for`, `class`...).
- Mayúsculas y minúsculas son distintas: `edad` ≠ `Edad`.

Palabras reservadas más comunes: `if`, `elif`, `else`, `for`, `while`, `def`, `return`, `import`, `class`, `True`, `False`, `None`, `and`, `or`, `not`, `in`, `is`, `break`, `continue`, `pass`.

### Convenciones (PEP 8)

| Qué | Estilo | Ejemplo |
|-----|--------|---------|
| Variables y funciones | `snake_case` | `precio_final`, `calcular_total()` |
| Constantes | `MAYUSCULAS` | `IVA = 0.16` |

- Usa nombres **descriptivos**: `edad_usuario` es mejor que `e`.
- Evita nombres que ocultan funciones que Python ya trae, como `print`, `type` o `round`.

```python
# Mal
n = 3
p = 12.5
t = n * p

# Bien
cantidad_productos = 3
precio_unitario = 12.5
total_compra = cantidad_productos * precio_unitario
total_compra
```

## Resumen

- `#` para comentarios que expliquen el *por qué*.
- Nombres en `snake_case`, descriptivos; constantes en `MAYUSCULAS`.
