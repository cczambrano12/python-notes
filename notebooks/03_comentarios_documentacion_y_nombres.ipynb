## Docstrings (documentación)

Un **docstring** es un texto entre triples comillas al inicio de una función. Describe qué hace, qué recibe y qué devuelve.

```python
def area_rectangulo(base, altura):
    """Calcula el área de un rectángulo.

    Parámetros:
        base: largo de la base.
        altura: largo de la altura.

    Devuelve:
        El área como número.
    """
    return base * altura

print(area_rectangulo(3, 4))
```

Python usa ese texto en su sistema de ayuda:

```python
help(area_rectangulo)
```

```python
print(area_rectangulo.__doc__)
```

## Nombres de variables

### Reglas obligatorias
- Pueden contener letras, números y `_`.
- No pueden empezar con número.
- No pueden tener espacios ni símbolos como `-` o `$`.
- No pueden ser palabras reservadas (`if`, `for`, `class`...).
- Mayúsculas y minúsculas son distintas: `edad` ≠ `Edad`.

```python
import keyword
print(keyword.kwlist)   # palabras reservadas
```

### Convenciones (PEP 8)

| Qué | Estilo | Ejemplo |
|-----|--------|---------|
| Variables y funciones | `snake_case` | `precio_final`, `calcular_total()` |
| Constantes | `MAYUSCULAS` | `IVA = 0.16` |
| Clases (más adelante) | `PascalCase` | `CuentaBancaria` |

- Usa nombres **descriptivos**: `edad_usuario` es mejor que `e`.
- Evita nombres que ocultan funciones de Python: `list`, `str`, `sum`, `max`.

```python
# Mal
n = 3
p = 12.5
t = n * p

# Bien
cantidad_productos = 3
precio_unitario = 12.5
total_compra = cantidad_productos * precio_unitario
print(total_compra)
```

## ¿Es un nombre válido?

Escribe un nombre y compruébalo.

```python
import ipywidgets as widgets
import keyword

caja = widgets.Text(value="mi_variable", description="Nombre:")
salida = widgets.Output()

def validar(_=None):
    salida.clear_output()
    n = caja.value
    with salida:
        if keyword.iskeyword(n):
            print("❌ Es una palabra reservada")
        elif not n.isidentifier():
            print("❌ No válido (¿empieza con número, tiene espacios o símbolos?)")
        elif n != n.lower() and not n.isupper():
            print("⚠️ Válido, pero la convención es snake_case")
        else:
            print("✅ Válido y sigue la convención")

caja.observe(validar, names="value")
validar()
display(caja, salida)
```

## Resumen

- `#` para comentarios que expliquen el *por qué*.
- Docstrings (`"""..."""`) para documentar funciones.
- Nombres en `snake_case`, descriptivos; constantes en `MAYUSCULAS`.
