# 07 · Ciclos: `while` y `for`

Un **ciclo** repite un bloque de código.

## `while`: repetir mientras se cumpla una condición

```python
contador = 1
while contador <= 5:
    print("Vuelta", contador)
    contador += 1       # sin esto, ¡ciclo infinito!
print("Fin")
```

⚠️ Siempre asegúrate de que algo cambie para que la condición llegue a ser falsa.

### Acumuladores

```python
suma = 0
n = 1
while n <= 100:
    suma += n
    n += 1
print("Suma de 1 a 100:", suma)
```

## `for`: recorrer una secuencia

Recorre cada elemento de una cadena, lista, rango, etc.

```python
for letra in "Hola":
    print(letra)
```

```python
for fruta in ["manzana", "pera", "uva"]:
    print("Me gusta la", fruta)
```

### `range(inicio, fin, paso)`

Genera números. Incluye `inicio`, **excluye** `fin`.

```python
print(list(range(5)))          # 0..4
print(list(range(2, 8)))       # 2..7
print(list(range(0, 10, 3)))   # de 3 en 3
print(list(range(5, 0, -1)))   # cuenta regresiva
```

```python
# Tabla del 7
for i in range(1, 11):
    print(f"7 x {i:2} = {7 * i:2}")
```

### `enumerate`: posición y valor a la vez

```python
colores = ["rojo", "verde", "azul"]
for i, color in enumerate(colores):
    print(i, color)
```

### Recorrer por índice (strings y listas)

```python
palabra = "casa"
for i in range(len(palabra)):
    print(f"palabra[{i}] = {palabra[i]}")
```

## ¿`for` o `while`?

- `for`: sabes **sobre qué** o **cuántas veces** iterar.
- `while`: repites **hasta que** pase algo (no sabes cuántas veces).

## `break`, `continue` y `pass`

| Palabra | Efecto |
|---------|--------|
| `break` | sale del ciclo inmediatamente |
| `continue` | salta a la siguiente vuelta |
| `pass` | no hace nada (marcador de posición) |

```python
# break: buscar el primer múltiplo de 7 mayor que 50
for n in range(50, 100):
    if n % 7 == 0:
        print("Encontrado:", n)
        break
```

```python
# continue: imprimir solo impares
for n in range(1, 10):
    if n % 2 == 0:
        continue
    print(n)
```

```python
# pass: bloque vacío que se llenará después
for n in range(3):
    if n == 1:
        pass   # TODO: hacer algo aquí
    else:
        print(n)

def funcion_pendiente():
    pass   # sin esto, un bloque vacío da error
```

## Pedir datos al usuario con `input()`

`input()` pausa el programa y devuelve **siempre texto** (`str`); conviértelo con `int()` o `float()`.

```python
nombre = input("¿Cómo te llamas? ")
print(f"Hola, {nombre}")
```

### `while` para validar la entrada

Repite la pregunta hasta que la respuesta sea válida. El patrón `while True` + `break` es muy común.

```python
while True:
    texto = input("Escribe un número entre 1 y 10: ")
    if not texto.isdigit():
        print("Eso no es un número.")
        continue
    numero = int(texto)
    if 1 <= numero <= 10:
        break
    print("Fuera de rango.")
print("Elegiste", numero)
```

### Menú que se repite hasta salir

```python
while True:
    opcion = input("(s)aludar, (c)ontar hasta 3, (q)uit: ").lower()
    if opcion == "s":
        print("¡Hola!")
    elif opcion == "c":
        for i in range(1, 4):
            print(i)
    elif opcion == "q":
        print("Adiós")
        break
    else:
        pass   # opción desconocida: ignorar y volver a preguntar
```

## Ciclos anidados

```python
for fila in range(1, 5):
    for col in range(1, fila + 1):
        print("*", end="")
    print()
```

## 🎛️ Visualiza `range`

Cambia los valores y mira qué números genera.

```python
import ipywidgets as widgets

inicio = widgets.IntSlider(value=0, min=-10, max=20, description="inicio")
fin = widgets.IntSlider(value=10, min=-10, max=30, description="fin")
paso = widgets.IntSlider(value=2, min=-5, max=5, description="paso")
out = widgets.Output()

def mostrar(_=None):
    out.clear_output()
    with out:
        if paso.value == 0:
            print("El paso no puede ser 0")
        else:
            print(f"range({inicio.value}, {fin.value}, {paso.value}) →",
                  list(range(inicio.value, fin.value, paso.value)))

for w in (inicio, fin, paso):
    w.observe(mostrar, names="value")
mostrar()
display(inicio, fin, paso, out)
```

## Resumen

- `while condición:` repite mientras sea verdadera; `for x in secuencia:` recorre elementos.
- `range()`, `enumerate()` ayudan a iterar.
- `break` sale, `continue` salta, `pass` no hace nada.
- `input()` devuelve texto; combínalo con `while` para validar.
