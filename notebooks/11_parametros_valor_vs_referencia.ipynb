# 11 · Paso de parámetros: por valor vs. por referencia

¿Qué pasa con una variable cuando la pasas a una función? Depende de si el tipo es **inmutable** o **mutable**.

| Inmutables (no cambian) | Mutables (se pueden modificar) |
|-------------------------|--------------------------------|
| `int`, `float`, `str`, `bool`, `tuple` | `list`, `dict`, `set` |

## Variables son "etiquetas" a objetos

`id()` muestra la identidad (dirección) del objeto en memoria.

```python
a = [1, 2, 3]
b = a            # b apunta al MISMO objeto
print(id(a) == id(b))

b.append(4)
print(a)         # ¡a también cambió!
```

## Inmutables: la función no puede cambiar el original

Se comporta como "por valor".

```python
def incrementar(n):
    n = n + 1          # crea un nuevo entero local
    print("dentro:", n)

x = 10
incrementar(x)
print("fuera:", x)    # sigue siendo 10
```

## Mutables: la función SÍ cambia el original

Se comporta como "por referencia".

```python
def agregar_elemento(lista):
    lista.append(99)
    print("dentro:", lista)

numeros = [1, 2, 3]
agregar_elemento(numeros)
print("fuera:", numeros)   # cambió
```

### Reasignar ≠ modificar

```python
def reasignar(lista):
    lista = [0, 0, 0]   # nueva lista local; el original no se toca

def modificar(lista):
    lista[0] = 0        # modifica el objeto original

datos = [1, 2, 3]
reasignar(datos)
print(datos)
modificar(datos)
print(datos)
```

## Cómo evitar modificar el original: copiar

```python
original = [1, 2, 3]

copia1 = original.copy()
copia2 = original[:]
copia3 = list(original)

copia1.append(4)
print(original, copia1)
```

### Copia superficial vs. profunda

`copy()` copia solo el primer nivel; los objetos internos se **comparten**. `copy.deepcopy()` copia todo.

```python
import copy

matriz = [[1, 2], [3, 4]]
superficial = matriz.copy()
profunda = copy.deepcopy(matriz)

matriz[0][0] = 99
print("superficial:", superficial)   # ¡también cambió!
print("profunda:   ", profunda)      # intacta
```

## Trampa clásica: valor por defecto mutable

```python
def agregar_mal(x, lista=[]):     # ¡MAL! la lista se crea una sola vez
    lista.append(x)
    return lista

print(agregar_mal(1))
print(agregar_mal(2))   # recuerda el 1

def agregar_bien(x, lista=None):
    if lista is None:
        lista = []
    lista.append(x)
    return lista

print(agregar_bien(1))
print(agregar_bien(2))
```

## `==` vs `is`

- `==`: ¿tienen el mismo **contenido**?
- `is`: ¿son el **mismo objeto**?

```python
a = [1, 2]
b = [1, 2]
c = a
print(a == b, a is b)
print(a == c, a is c)
```

## Resumen

- Inmutables (`int`, `str`, `tuple`...): la función no los puede alterar.
- Mutables (`list`, `dict`): la función los modifica en el sitio.
- Para proteger el original: `.copy()` o `copy.deepcopy()`.
- Nunca uses `[]` o `{}` como valor por defecto.
